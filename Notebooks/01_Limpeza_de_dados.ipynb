## 📋 Tabela de Produtos — (olist_products)

Contém informações cadastrais dos produtos como categoria, dimensões e peso.

### 🔍 Tipos de Dados

```python
produtos.dtypes
```

```python
produtos.head(1)
```

### 🔍 Valores Nulos

```python
produtos.isnull().sum()
```

### 🛠️ Tratamento de Valores Nulos

- 610 nulos em `product_category_name` — produtos sem categoria cadastrada
- 610 nulos em atributos descritivos do produto
- 2 nulos em peso e dimensões físicas — registros incompletos

**Estratégia:**

- `product_category_name` → preencher com `'Sem Categoria'`
- Atributos numéricos → preencher com a mediana da respectiva coluna

A mediana foi escolhida para os atributos numéricos por ser menos influenciada por valores extremos (outliers), podendo representar melhor o valor central da distribuição.

```python
# Preenchendo categoria nula
produtos['product_category_name'] = produtos['product_category_name'].fillna('Sem Categoria')

# Preenchendo dimensões físicas com mediana

colunas_medianas = [
     'product_name_lenght',
    'product_description_lenght', 
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]

for coluna in colunas_medianas:
    produtos[coluna] = produtos[coluna].fillna(produtos[coluna].median())

#COnfirmando
produtos.isnull().sum()
```

### converte product_photos_qty para inteiro

```python
produtos['product_photos_qty'] = produtos['product_photos_qty'].astype(int)

produtos.dtypes
```
